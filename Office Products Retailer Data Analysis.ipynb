```python
import pandas as pd

import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import plotly.colors as colors

pio.templates.default='plotly_white'
```

```python
data=pd.read_csv('Sample - Superstore.csv', encoding = 'latin-1')

#Latin-1 is a character encoding standard officially known as ISO 8859-1

data
```

```python
data.describe()
```

```python
data.info()
```

<h4>Converting Ship Date and Order Date to datetime format from object type</h4>

```python
data['Order Date']=pd.to_datetime(data['Order Date'])
data['Ship Date']=pd.to_datetime(data['Ship Date'])
data.info()
```

<h4>Creating columns to calculate and store data for profits in a year, month and week </h4>

```python
data['Order Year']= data['Order Date'].dt.year
data['Order Month']= data['Order Date'].dt.month_name()
data['Order Day of Week']= data['Order Date'].dt.dayofweek

data.head()
```

<h3>Monthly Sales Analysis</h3>

```python
sales_by_month=data.groupby('Order Month')['Sales'].sum().reset_index()
sales_by_month         
```

```python
monthsales_Chart=px.line(sales_by_month,
                         x='Order Month',
                         y='Sales',
                         title="Monthly Sales Analysis")
monthsales_Chart.show()
```

```python
pio.write_html(monthsales_Chart, file="monthlysales.html", auto_open= True)
```

<h3>Sales by category</h3>

```python
sales_by_category=data.groupby('Category')['Sales'].sum().reset_index()
sales_by_category
```

```python
categorysales_Chart=px.pie(sales_by_category,
                        values="Sales",
                           names="Category",
                           hole=0.5,
                           color_discrete_sequence=px.colors.qualitative.Pastel)
categorysales_Chart.update_traces(textposition='inside', textinfo='percent+label')
categorysales_Chart.update_layout(title_text='Sales Analysis by Category',title_font=dict(size=24))
categorysales_Chart.show()
```

```python
pio.write_html(categorysales_Chart, file="categorysales.html", auto_open= True)
```

<h3>Sales Analysis by Sub-Category</h3>

```python
sales_by_subcategory=data.groupby('Sub-Category')['Sales'].sum().reset_index()
sales_by_subcategory
```

```python
subcategorysales_Chart=px.bar(sales_by_subcategory, x='Sub-Category', y='Sales', title='Sales Analysis by Sub-Category')
subcategorysales_Chart.show()
```

```python
pio.write_html(subcategorysales_Chart, file="subcategorysales.html", auto_open= True)
```

<h3>Sales and Profit by Customer Segment</h3>

```python
sales_profit_by_segment=data.groupby('Segment').agg({'Sales':'sum','Profit':'sum'}).reset_index()
sales_profit_by_segment
```

```python
color_palette=colors.qualitative.Pastel

segmentsalesprofit_Chart=go.Figure()
segmentsalesprofit_Chart.add_trace(go.Bar(x=sales_profit_by_segment['Segment'],
                                          y=sales_profit_by_segment['Profit'],
                                          name='Profit',
                                          marker_color=color_palette[0]))
segmentsalesprofit_Chart.add_trace(go.Bar(x=sales_profit_by_segment['Segment'],
                                          y=sales_profit_by_segment['Sales'],
                                          name='Sales',
                                          marker_color=color_palette[1]))
segmentsalesprofit_Chart.update_layout(title='Sales and Profit Analysis by Customer Segment',
                                       xaxis_title='Customer Segment',
                                       yaxis_title='Amount')
segmentsalesprofit_Chart.show()                                       
```

```python
pio.write_html(segmentsalesprofit_Chart, file="segmentsalesprofit.html", auto_open= True)
```

<h4>Sales to Profit Ratio for Customer Segment</h4>

```python
sales_profit_by_segment=data.groupby('Segment').agg({'Sales':'sum','Profit':'sum'}).reset_index()
sales_profit_by_segment['Sales to Profit Ratio']=sales_profit_by_segment['Sales']/sales_profit_by_segment['Profit']
print(sales_profit_by_segment[['Segment','Sales to Profit Ratio']])
```
